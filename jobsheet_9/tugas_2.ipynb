- Buatlah model klasfikasi Multinomial Naive Bayes dengan ketentuan,
  1. Menggunakan data `spam.csv`
  2. Fitur `CountVectorizer` dengan mengaktifkan **stop_words**
  3. Evaluasi hasilnya



```python
import numpy as np
import pandas as pd

df = pd.read_csv('spam.csv', encoding='latin-1') # spesifiksi encoding diperlukan karena data tidak menggunakan UTF-8

df.head()
```

```python
# Drop 3 kolom terakhir dengan fungsi iloc
df = df.drop(df.iloc[:,2:], axis=1)

# Ubah nama kolom v1 dan v2
df = df.rename(columns={'v1': 'Labels', 'v2': 'SMS'})

# Cek data
df.head()
```

```python
# Cek Jumlah Data Per Kelas
print(df['Labels'].value_counts())
print('\n')

# Cek Kelengkapan Data
print(df.info())
print('\n')

# Cek Statistik Deskriptif
print(df.describe())
```

```python
# Data untuk label
new_labels = {
    'spam': 1,
    'ham': 0
}

# Encode label
df['Labels'] = df['Labels'].map(new_labels)

# Cek data
df.head()
```

```python
X = df['SMS'].values
y = df['Labels'].values
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# Memisahan Data Train dan Test (Teks Mentah)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=50)

# Inisialisasi CountVectorizer dengan mengaktifkan stop_words
bow_tugas1 = CountVectorizer(stop_words='english')

# Transformasi Teks ke Matriks Ekstraksi Fitur
X_train_bow = bow_tugas1.fit_transform(X_train)
X_test_bow = bow_tugas1.transform(X_test)

# Inisialisasi dan Pelatihan Model Multinomial Naive Bayes
mnb_tugas1 = MultinomialNB()
mnb_tugas1.fit(X_train_bow, y_train)
```

```python
# Evaluasi Hasil Prediksi
y_pred_train_tugas1 = mnb_tugas1.predict(X_train_bow)
y_pred_test_tugas1 = mnb_tugas1.predict(X_test_bow)

acc_train_tugas1 = accuracy_score(y_train, y_pred_train_tugas1)
acc_test_tugas1 = accuracy_score(y_test, y_pred_test_tugas1)

# Tampilan Output
print(f'Jumlah fitur setelah stop words: {len(bow_tugas1.get_feature_names_out())}')
print(f'Dimensi data train: {X_train_bow.shape}')
print(f'Akurasi Data Train (Tugas 1): {acc_train_tugas1}')
print(f'Akurasi Data Test (Tugas 1): {acc_test_tugas1}')
```

- Buatlah model klasfikasi Multinomial Naive Bayes dengan ketentuan,
  1. Menggunakan data `spam.csv`
  2. Fitur `TF-IDF` dengan mengaktifkan **stop_words**
  3. Evaluasi hasilnya dan bandingkan dengan hasil pada Tugas no 2.
  4. Berikan kesimpulan fitur mana yang terbaik pada kasus data `spam.csv`

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# Inisialisasi TfidfVectorizer dengan stop_words='english'
tfidf = TfidfVectorizer(stop_words='english')

# Transformasi data teks mentah ke matriks TF-IDF
X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

# Inisialisasi dan latih model Multinomial Naive Bayes
mnb_tfidf = MultinomialNB()
mnb_tfidf.fit(X_train_tfidf, y_train)

# Prediksi dan Evaluasi Model
y_pred_train_tfidf = mnb_tfidf.predict(X_train_tfidf)
y_pred_test_tfidf = mnb_tfidf.predict(X_test_tfidf)

acc_train_tfidf = accuracy_score(y_train, y_pred_train_tfidf)
acc_test_tfidf = accuracy_score(y_test, y_pred_test_tfidf)

# Output Hasil
print(f'Jumlah fitur (TF-IDF): {len(tfidf.get_feature_names_out())}')
print(f'Dimensi data train: {X_train_tfidf.shape}')
print(f'Akurasi Data Train (TF-IDF): {acc_train_tfidf}')
print(f'Akurasi Data Test (TF-IDF): {acc_test_tfidf}')
```

### Perbandingan dan Kesimpulan

#### Perbandingan Hasil Evaluasi
* **CountVectorizer (Tugas 1)**:
  * Akurasi Train: 99.46%
  * Akurasi Test: 98.30%
* **TfidfVectorizer (Tugas 2)**:
  * Akurasi Train: ~98.61%
  * Akurasi Test: ~96.68%

#### Kesimpulan Fitur Terbaik
Fitur **CountVectorizer (Bag of Words)** memberikan performa yang **lebih baik** dibandingkan `TfidfVectorizer` pada kasus dataset `spam.csv`.

**Alasan Utama**:
1. Pola Kemunculan Kata Kunci Spam<br>
Pada klasifikasi pesan SMS spam, kemunculan kata-kata pemicu tertentu (seperti *FREE*, *WIN*, *CLAIM*, *URGENT*, atau digit angka) dalam jumlah/frekuensi langsung lebih terdeteksi secara optimal oleh perhitungan frekuensi murni (`CountVectorizer`).
2. Pengaruh Penimbangan Bobot TF-IDF<br>
`TfidfVectorizer` memberikan bobot penalti pada kata-kata yang terlalu sering muncul di banyak dokumen. Pada pesan SMS pendek, penimbangan pembagi IDF ini justru sedikit menurunkan sensitivitas Multinomial Naive Bayes terhadap kata-kata khas yang sering berulang pada pesan spam.
