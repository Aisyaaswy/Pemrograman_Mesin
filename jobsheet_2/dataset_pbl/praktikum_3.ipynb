# **PRAKTIKUM 3**


```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
```

# **Langkah 1 - Load Data**

```python
dpath = 'Sleep_health_and_lifestyle_dataset.csv'
df = pd.read_csv(dpath)
df.head()
```

# **Langkah 2 - Slice Data**


```python
df = df[['Quality of Sleep', 'Gender', 'Age', 'Occupation', 'Sleep Duration', 'Stress Level', 'BMI Category']]
df.head()
```

#**Langkah 3 - Encoding**


```python
le = LabelEncoder() # membuat objek dari LabelEncoder
df['Gender'] = le.fit_transform(df['Gender']) # proses encoding
df['Occupation'] = le.fit_transform(df['Occupation']) # proses encoding
df['BMI Category'] = le.fit_transform(df['BMI Category']) # proses encoding
```

# **Langkah 4 - Verifikasi Hasil**

```python
df.head()
```

# **Langkah 5 - Standarisasi**

```python
std = StandardScaler()
df['Age'] = std.fit_transform(df[['Age']])
```
