# **Tugas Praktikum**
Memisahkan antara variabel yang dapat digunakan dan variabel yang tidak dapat digunakan, melakukan proses encoding pada kolom kategorikal dan target "Sleep Disorder", serta melakukan proses standardisasi pada semua kolom yang memiliki nilai numerik dengan menggunakan data "Sleep Health and Lifestyle" yang terdiri dari 374 data untuk memprediksi gangguan tidur (Insomnia, Sleep Apnea, dan None).

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
df = pd.read_csv('Sleep_health_and_lifestyle_dataset.csv')
df.head()
```

```python
# Memisahkan Variabel yang Digunakan dan Tidak Digunakan
df = df.drop(columns=['Person ID', 'Blood Pressure'])
```

```python
# Mengisi nilai kosong pada Sleep Disorder (jika terbaca NaN) agar aman saat di-encode
df['Sleep Disorder'] = df['Sleep Disorder'].fillna('None')
```

```python
# Encoding pada Kolom kategorikal & target diagnosis ('Sleep Disorder')
le = LabelEncoder()
kolom_kategorikal = ['Gender', 'Occupation', 'BMI Category', 'Sleep Disorder']
for col in kolom_kategorikal:
    df[col] = le.fit_transform(df[col])
```

```python
# Standardisasi pada Semua Kolom Fitur Numerik
kolom_numerik = ['Age', 'Sleep Duration', 'Quality of Sleep', 'Physical Activity Level', 'Stress Level', 'Heart Rate', 'Daily Steps']
scaler = StandardScaler()
df[kolom_numerik] = scaler.fit_transform(df[kolom_numerik])
```

```python
# Verifikasi Hasil Pra-pemrosesan
print("Dimensi dataset setelah diproses:", df.shape)
df.head()
```
