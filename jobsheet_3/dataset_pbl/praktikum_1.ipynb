#**Langkah 3 - Imputasi**


```python
# Mengisi nilai kosong pada kolom Sleep Disorder dengan kategori 'None' (Tidak ada gangguan)
df['Sleep Disorder'] = df['Sleep Disorder'].fillna('None')
```

# **Langkah 5 - Ekstraksi Fitur**
Setelah melakukan Pre-Processing, langkah berikutnya kita akan melakukan Ekstraksi Fitur berdasarkan Jumlah keluarga "Familysize", Gelar Jumlah keluarga "Title", kelompok usia "AgeBin", Tempat tiket "CabinDeck"

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# === EKSTRAKSI FITUR DATASET SLEEP HEALTH ===

# 1. AgeBin: mengelompokkan usia menjadi kategori (YoungAdult, Adult, MiddleAge, Senior)
df["AgeBin"] = pd.cut(df["Age"], bins=[18, 30, 45, 60, 100],
                      labels=["YoungAdult", "Adult", "MiddleAge", "Senior"])

# 2. Memecah kolom Blood Pressure ('126/83') menjadi dua kolom angka: Systolic dan Diastolic
bp_split = df["Blood Pressure"].str.split("/", expand=True)
df["Systolic_BP"] = bp_split[0].astype(int)
df["Diastolic_BP"] = bp_split[1].astype(int)

# 3. SleepEfficiencyScore: kombinasi rasio durasi tidur dengan kualitas tidur
df["Sleep_Index"] = (df["Sleep Duration"] * df["Quality of Sleep"]).round(2)

# 4. Activity_Stress_Ratio: perbandingan tingkat aktivitas fisik terhadap tingkat stres
df["Activity_Stress_Ratio"] = (df["Physical Activity Level"] / df["Stress Level"]).round(2)

# 5. Merapikan kategori BMI ('Normal Weight' digabung ke 'Normal')
df["BMI Category"] = df["BMI Category"].replace({"Normal Weight": "Normal"})

print("✅ Hasil Ekstraksi Fitur")
df[["Age", "AgeBin", "Blood Pressure", "Systolic_BP", "Diastolic_BP", "Sleep_Index", "Activity_Stress_Ratio", "BMI Category"]].head()
```

# **Langkah 6 - Visualisasi Hasil Ekstraksi Fitur**


```python
import seaborn as sns
import matplotlib.pyplot as plt

# === VISUALISASI FITUR HASIL EKSTRAKSI ===
plt.figure(figsize=(10, 5))
sns.barplot(x="AgeBin", y="Quality of Sleep", data=df)
plt.title("Rata-rata Kualitas Tidur berdasarkan Kelompok Usia (AgeBin)")
plt.show()

plt.figure(figsize=(10, 5))
sns.barplot(x="Sleep Disorder", y="Stress Level", data=df)
plt.title("Rata-rata Tingkat Stres berdasarkan Gangguan Tidur")
plt.show()

plt.figure(figsize=(10, 5))
sns.barplot(x="BMI Category", y="Sleep Duration", data=df)
plt.title("Rata-rata Durasi Tidur berdasarkan Kategori BMI")
plt.show()
```
