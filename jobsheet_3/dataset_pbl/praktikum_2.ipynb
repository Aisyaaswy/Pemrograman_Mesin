# **PRAKTIKUM 2**
# **Langkah 0 - Load Library**

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

# **Langkah 1 - Load Data**

```python
dpath = 'Sleep_health_and_lifestyle_dataset.csv'
df = pd.read_csv(dpath)
df.head()
```

# **Langkah 2 - Pre Processing Encoding Sex**

```python
# --- 2. Encoding fitur 'Sex' ---
df['Gender'] = df['Gender'].map({'male': 1, 'female': 0})
```

# **Langkah 3 - Verifikasi hasil encoding 'Sex'**


```python
df['Gender'] = df['Gender'].map({'Male': 1, 'Female': 0})
print("=== Hasil Setelah Encoding 'Gender' ===")
print(df[['Gender']].head(), "\n")
```

#**Langkah 4 - Pre Processing Encoding Fitur Kategori**


```python
# --- 3. Encoding fitur 'BMI Category', 'Occupation', & 'Sleep Disorder' ---
df['Sleep Disorder'] = df['Sleep Disorder'].fillna('None')

le = LabelEncoder()
df['BMI Category'] = le.fit_transform(df['BMI Category'])
df['Occupation'] = le.fit_transform(df['Occupation'])
df['Sleep Disorder'] = le.fit_transform(df['Sleep Disorder'])
```

#**Langkah 5 - Verifikasi hasil encoding Fitur Kategori**
## **Pre Processing Imputasi Data**

```python
# --- 4. Imputasi fitur 'Age' ---
df['Age'] = df['Age'].fillna(df['Age'].median())
print("=== Hasil Setelah Imputasi Data pada Usia dengan median ===")
print(df[['Age']].head(10), "\n")
```

# **Langkah 6 - Pre Processing Standarisasi**

```python
# --- 5. Standarisasi fitur 'Age' ---
scaler = StandardScaler()
df['Age'] = scaler.fit_transform(df[['Age']])
print("=== Hasil Setelah Standardisasi 'Age' ===")
print(df[['Age']].head(10), "\n")
```

# **Langkah 7 - Hasil dan Verifikasi setelah Pre Processing**

```python
# --- 6. Cek hasil akhir ---
print("=== Hasil Preprocessing (5 baris pertama) ===")
df.head()
```

# **Langkah 8 - Seleksi Fitur**

```python
df = df[['Quality of Sleep', 'Gender', 'Age', 'Occupation', 'Sleep Duration', 'Physical Activity Level', 'Stress Level', 'BMI Category', 'Daily Steps', 'Sleep Disorder']]
df.head()
```
