# **Random Split**
## **Langkah 1 - Load Data**

```python
import pandas as pd

df = pd.read_csv('Sleep_health_and_lifestyle_dataset.csv')
df.head()
```

## **Langkah 2 - Split Data**


```python
# Split data
from sklearn.model_selection import train_test_split

# Split data training dan lainnya
# Rasio yang digunakan adalah 8:1:1
df_train, df_unseen = train_test_split(df, test_size=0.2, random_state=0)

# Split lagi antara validasi dan testing
df_val, df_test = train_test_split(df_unseen, test_size=0.5, random_state=0)

# Cek masing-masing ukuran data
print(f'Jumlah data asli: {df.shape[0]}')
print(f'Jumlah data train: {df_train.shape[0]}')
print(f'Jumlah data val: {df_val.shape[0]}')
print(f'Jumlah data test: {df_test.shape[0]}')

# Cek rasio tiap label
print('=========')
print(f'Jumlah label data asli:\n{df["Quality of Sleep"].value_counts()}')
print(f'Jumlah label data train:\n{df_train["Quality of Sleep"].value_counts()}')
print(f'Jumlah label data val:\n{df_val["Quality of Sleep"].value_counts()}')
print(f'Jumlah label data test:\n{df_test["Quality of Sleep"].value_counts()}')
```

#**Stratified Split**
## **Langkah 1 - Load Data**


```python
import pandas as pd

df2 = pd.read_csv('Sleep_health_and_lifestyle_dataset.csv')
df2.head()
```

## **Langkah 2 - Split Data**

```python
from sklearn.model_selection import train_test_split

# Gunakan test_size=0.4 agar kelas minoritas punya minimal 2 data di df2_unseen
df2_train, df2_unseen = train_test_split(df2, test_size=0.4, random_state=0, stratify=df2['Quality of Sleep'])

# Split lagi antara validasi dan testing dengan stratify
df2_val, df2_test = train_test_split(df2_unseen, test_size=0.5, random_state=0, stratify=df2_unseen['Quality of Sleep'])

# Cek masing-masing ukuran data
print(f'Jumlah label data asli:\n{df2["Quality of Sleep"].value_counts()}\n')
print(f'Jumlah label data train:\n{df2_train["Quality of Sleep"].value_counts()}\n')
print(f'Jumlah label data val:\n{df2_val["Quality of Sleep"].value_counts()}\n')
print(f'Jumlah label data test:\n{df2_test["Quality of Sleep"].value_counts()}')
```

# **Cross Validation 1**
## **Langkah 1 - Load Data**

```python
import pandas as pd

df3 = pd.read_csv('Sleep_health_and_lifestyle_dataset.csv')
df3.head()
```

## **Langkah 2 - Split Data**

```python
# Implementasi k-fold cross validation (random) dengan training dan testing saja
from sklearn.model_selection import KFold

# Inisiasi obyek kfold
kf = KFold(n_splits=4)
print(f'Jumlah fold: {kf.get_n_splits()}')
print(f'Obyek KFold: {kf}')

# Lakukan splitting dengan KFold
kf_split = kf.split(df3)
print(f'Jumlah data df3: {df3.shape[0]}')

# Cek index data tiap fold
for train_index, test_index in kf_split:
    print(f'Index train: {train_index} | Index test: {test_index}')
```

# **Cross Validation 2**
## **Langkah 1 - Load Data**

```python
import pandas as pd

df4 = pd.read_csv('Sleep_health_and_lifestyle_dataset.csv')
df4.head()
```

## **Langkah 2 - Split Data**

```python
# Implementasi k-fold cross validation (random) dengan training, validation, dan testing data
from sklearn.model_selection import train_test_split, KFold

# Split dulu antara data train dan test (rasio 8:2)
df4_train, df4_test = train_test_split(df4, test_size=0.2, random_state=0)

# Inisiasi obyek kfold
kf2 = KFold(n_splits=4)
print(f'Jumlah fold: {kf2.get_n_splits()}')
print(f'Obyek KFold: {kf2}')

# Lakukan splitting dengan KFold pada data train
kf2_split = kf2.split(df4_train)
print(f'Jumlah data df4_train: {df4_train.shape[0]}')

# Cek index data tiap fold
for train_index, test_index in kf2_split:
    print(f'Index train: {train_index} | Index val: {test_index}')
```
