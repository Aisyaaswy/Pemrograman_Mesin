# **PRAKTIKUM 1**

Praktikum 1 ini akan membuat JST sederhana (2 layer) dengan forward pass dan backpropagation manual.

**Langkah:**
1. Buat dataset sederhana (XOR).
2. Inisialisasi bobot dan bias.
3. Implementasikan forward pass.
4. Hitung error dan lakukan backpropagation.
5. Update bobot menggunakan gradient descent.

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# 1. Load dataset Sleep Health
df = pd.read_csv('Sleep_health_and_lifestyle_dataset.csv')

fitur = [
    'Age', 'Sleep Duration', 'Quality of Sleep',
    'Physical Activity Level', 'Stress Level',
    'Heart Rate', 'Daily Steps'
]
X_raw = df[fitur].values
y_raw = df['Sleep Disorder'].apply(
    lambda x: 0 if pd.isna(x) or x == 'None' else 1
).values.reshape(-1, 1)

# Bagi data dan lakukan normalisasi
X_train, X_test, y_train, y_test = train_test_split(
    X_raw, y_raw, test_size=0.2, random_state=42
)

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# 2. Parameter jaringan
np.random.seed(42)
input_size = X_train.shape[1]
hidden_size = 8
output_size = 1
lr = 0.01

# Inisialisasi bobot dan bias
W1 = np.random.randn(input_size, hidden_size) * 0.1
b1 = np.zeros((1, hidden_size))
W2 = np.random.randn(hidden_size, output_size) * 0.1
b2 = np.zeros((1, output_size))

# Fungsi aktivasi Sigmoid
def sigmoid(x):
    return 1 / (1 + np.exp(-np.clip(x, -500, 500)))

def sigmoid_derivative(x):
    return x * (1 - x)

# 3. Training
for epoch in range(10000):
    # Forward pass
    z1 = np.dot(X_train, W1) + b1
    a1 = sigmoid(z1)
    z2 = np.dot(a1, W2) + b2
    a2 = sigmoid(z2)

    # Hitung error
    error = y_train - a2

    # Backpropagation
    d_a2 = error * sigmoid_derivative(a2)
    d_W2 = np.dot(a1.T, d_a2)
    d_b2 = np.sum(d_a2, axis=0, keepdims=True)

    d_a1 = np.dot(d_a2, W2.T) * sigmoid_derivative(a1)
    d_W1 = np.dot(X_train.T, d_a1)
    d_b1 = np.sum(d_a1, axis=0, keepdims=True)

    # Update bobot
    W1 += lr * d_W1
    b1 += lr * d_b1
    W2 += lr * d_W2
    b2 += lr * d_b2

    if epoch % 1000 == 0:
        loss = np.mean(np.square(error))
        print(f"Epoch {epoch}, Loss: {loss:.4f}")

# 4. Evaluasi pada data test
a1_test = sigmoid(np.dot(X_test, W1) + b1)
a2_test = sigmoid(np.dot(a1_test, W2) + b2)

prediksi = (a2_test >= 0.5).astype(int)
akurasi = np.mean(prediksi == y_test)
print(f"\nAkurasi Data Test (Sigmoid): {akurasi * 100:.2f}%")
```

## **TUGAS 1:**
1. Ubah jumlah neuron hidden layer menjadi 3.
2. Bandingkan hasil loss dengan konfigurasi awal.
3. Tambahkan fungsi aktivasi ReLU dan bandingkan hasil.

```python
# Fungsi aktivasi ReLU untuk hidden layer
def relu(x):
    return np.maximum(0, x)

def relu_derivative(z):
    return np.where(z > 0, 1, 0)

# Inisialisasi ulang parameter untuk Tugas 1 (Hidden Layer ReLU)
np.random.seed(42)
hidden_size_tugas = 16
lr_tugas = 0.005

W1_t = np.random.randn(input_size, hidden_size_tugas) * 0.1
b1_t = np.zeros((1, hidden_size_tugas))
W2_t = np.random.randn(hidden_size_tugas, output_size) * 0.1
b2_t = np.zeros((1, output_size))

# Training dengan ReLU di hidden layer
for epoch in range(10000):
    # Forward pass
    z1 = np.dot(X_train, W1_t) + b1_t
    a1 = relu(z1)
    z2 = np.dot(a1, W2_t) + b2_t
    a2 = sigmoid(z2)

    # Hitung error
    error = y_train - a2

    # Backpropagation
    d_a2 = error * sigmoid_derivative(a2)
    d_W2 = np.dot(a1.T, d_a2)
    d_b2 = np.sum(d_a2, axis=0, keepdims=True)

    d_a1 = np.dot(d_a2, W2_t.T) * relu_derivative(z1)
    d_W1 = np.dot(X_train.T, d_a1)
    d_b1 = np.sum(d_a1, axis=0, keepdims=True)

    # Update bobot
    W1_t += lr_tugas * d_W1
    b1_t += lr_tugas * d_b1
    W2_t += lr_tugas * d_W2
    b2_t += lr_tugas * d_b2

    if epoch % 1000 == 0:
        loss = np.mean(np.square(error))
        print(f"[Tugas 1 - ReLU] Epoch {epoch}, Loss: {loss:.4f}")

# Evaluasi pada data test
a1_test_t = relu(np.dot(X_test, W1_t) + b1_t)
a2_test_t = sigmoid(np.dot(a1_test_t, W2_t) + b2_t)

prediksi_t = (a2_test_t >= 0.5).astype(int)
akurasi_t = np.mean(prediksi_t == y_test)
print(f"\nAkurasi Data Test (ReLU): {akurasi_t * 100:.2f}%")
```

- Perbandingan hidden_size:<br>
 Menambah jumlah neuron pada hidden layer dari 8 menjadi 16 meningkatkan kapasitas jaringan saraf dalam menangkap pola hubungan non-linear antar fitur dataset Sleep Health. Dengan neuron lebih banyak, model lebih leluasa membentuk kombinasi fitur (seperti interaksi antara durasi tidur, tingkat stres, dan detak jantung), sehingga penurunan nilai loss lebih stabil saat pelatihan. Namun, pada dataset berukuran kecil seperti Sleep Health yang hanya berjumlah 374 baris, neuron yang terlalu banyak berisiko memperlambat komputasi dan memicu overfitting jika tidak diimbangi pengaturan learning rate.
- Perbandingan Sigmoid vs ReLU:<br>
  - Sigmoid: Mengubah nilai masukan ke rentang 0 hingga 1 dengan kurva melengkung yang halus. Pada jaringan dangkal, Sigmoid bekerja stabil karena menghasilkan gradien yang terkendali di setiap iterasi, meskipun proses penurunannya cenderung lebih lambat saat nilai z berada di area ekstrem (terlalu besar atau terlalu kecil).

   - ReLU: Meneruskan nilai positif apa adanya dan mengubah nilai negatif menjadi 0. Pada eksperimen dataset Sleep Health, ReLU memberikan konvergensi yang jauh lebih cepat dan mencapai akurasi uji hingga 96.00% dengan loss yang turun hingga 0.0525, karena gradien untuk nilai aktif tetap konstan bernilai 1 dan terhindar dari masalah penurunan gradien (vanishing gradient).
