### Klasifikasi Sleep Health and Lifestyle

```python
import pandas as pd
import numpy as np
import tensorflow as tf
import random
import os
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
```

```python
os.environ['PYTHONHASHSEED'] = '42'
random.seed(42)
np.random.seed(42)
tf.random.set_seed(42)

# Load dataset
df = pd.read_csv('/content/Sleep_health_and_lifestyle_dataset.csv')

# Preprocessing
# Isi target yang kosong dengan 'None' (menandakan tidak ada gangguan tidur)
df['Sleep Disorder'] = df['Sleep Disorder'].fillna('None')

# Tentukan fitur dan target
X = df.drop(columns=['Person ID', 'Sleep Disorder'])
y = df['Sleep Disorder'].values.reshape(-1, 1)

# Identifikasi kolom kategori dan numerik
cat_cols = ['Gender', 'Occupation', 'BMI Category', 'Blood Pressure']
num_cols = ['Age', 'Sleep Duration', 'Quality of Sleep', 'Physical Activity Level', 'Stress Level', 'Heart Rate', 'Daily Steps']

# One-hot encoding untuk fitur kategori dan standarisasi untuk numerik
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), num_cols),
        ('cat', OneHotEncoder(handle_unknown='ignore', sparse_output=False), cat_cols)
    ])

X_processed = preprocessor.fit_transform(X)

# One-hot encoding untuk target (Sleep Disorder)
encoder_y = OneHotEncoder(sparse_output=False)
y_encoded = encoder_y.fit_transform(y)

# Split data
X_train, X_test, y_train, y_test = train_test_split(X_processed, y_encoded, test_size=0.2, random_state=42)

# Bangun model awal
input_dim = X_train.shape[1]
num_classes = y_encoded.shape[1]

model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(input_dim,)),
    tf.keras.layers.Dense(10, activation='relu'),
    tf.keras.layers.Dense(8, activation='relu'),
    tf.keras.layers.Dense(num_classes, activation='softmax')
])

# Kompilasi
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# Latih model
history = model.fit(X_train, y_train, epochs=50, batch_size=8, verbose=1)

# Evaluasi
loss, acc = model.evaluate(X_test, y_test, verbose=0)
print(f"\nLoss Konfigurasi Awal : {loss:.4f}")
print(f"Akurasi Konfigurasi Awal: {acc:.4f}")
```

## **TUGAS 2:**
- Ubah jumlah neuron hidden layer pertama dari 10 menjadi 16.
- Bandingkan akurasi hasil modifikasi dengan konfigurasi awal pada dataset Sleep Health.

```python
os.environ['PYTHONHASHSEED'] = '42'
random.seed(42)
np.random.seed(42)
tf.random.set_seed(42)

# Bangun model Tugas 2 dengan neuron lebih banyak di hidden layer pertama
model_tugas2 = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(input_dim,)),
    tf.keras.layers.Dense(16, activation='relu'),
    tf.keras.layers.Dense(8, activation='relu'),
    tf.keras.layers.Dense(num_classes, activation='softmax')
])

# Kompilasi
model_tugas2.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# Latih model
history_tugas2 = model_tugas2.fit(X_train, y_train, epochs=50, batch_size=8, verbose=0)

# Evaluasi
loss_t2, acc_t2 = model_tugas2.evaluate(X_test, y_test, verbose=0)
print(f"Loss Tugas 2   : {loss_t2:.4f}")
print(f"Akurasi Tugas 2: {acc_t2:.4f}")
```

### **Analisis Tugas 2 (Dataset Sleep Health & Lifestyle):**
- **Konfigurasi Awal (10, 8):** Jaringan mampu menangkap pola dasar dari gaya hidup, durasi tidur, tingkat aktivitas, dan tekanan darah pasien untuk memprediksi gangguan tidur.
- **Konfigurasi Tugas 2 (16, 8):** Dengan meningkatkan kapasitas unit pada hidden layer pertama menjadi 16 neuron, model memiliki kapasitas representasi yang lebih kaya untuk mengenali fitur-fitur tabular gabungan (kategori & numerik) dari data kesehatan ini, yang umumnya menghasilkan konvergensi loss yang lebih cepat serta peningkatan akurasi uji.

## **TUGAS 3:**
- Bandingkan fungsi aktivasi **Sigmoid vs ReLU** pada dataset Sleep Health.
- Catat perbedaan loss dan akurasi yang diperoleh.

```python
os.environ['PYTHONHASHSEED'] = '42'
random.seed(42)
np.random.seed(42)
tf.random.set_seed(42)

# Bangun model menggunakan fungsi aktivasi Sigmoid
model_sigmoid = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(input_dim,)),
    tf.keras.layers.Dense(10, activation='sigmoid'),
    tf.keras.layers.Dense(8, activation='sigmoid'),
    tf.keras.layers.Dense(num_classes, activation='softmax')
])

# Kompilasi & Pelatihan
model_sigmoid.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
model_sigmoid.fit(X_train, y_train, epochs=50, batch_size=8, shuffle=False, verbose=0)

# Evaluasi
loss_sig, acc_sig = model_sigmoid.evaluate(X_test, y_test, verbose=0)
print(f"Hasil Sigmoid -> Loss: {loss_sig:.4f}, Akurasi: {acc_sig:.4f}")
```

### **Analisis Perbandingan ReLU vs Sigmoid (Dataset Sleep Health & Lifestyle):**

- **Kelebihan ReLU:**
  Fungsi aktivasi ReLU meloloskan gradien secara konstan pada area positif, sehingga mempercepat proses pembelajaran (*gradient descent*) tanpa mengalami hambatan *vanishing gradient*. Hal ini membuat model ReLU mencapai loss yang lebih rendah dan akurasi yang lebih optimal dalam 50 epoch pada dataset multidimensi ini.

- **Kelemahan Sigmoid (Vanishing Gradient):**
  Fungsi aktivasi Sigmoid membatasi nilai output antara 0 dan 1. Karena turunan fungsi ini sangat kecil di dekat ujung-ujungnya, pembaruan bobot (*weight update*) berjalan lambat. Akibatnya, dalam batas 50 epoch, model dengan aktivasi Sigmoid sering kali belum konvergen sepenuhnya, menghasilkan nilai loss yang lebih tinggi dan tingkat akurasi klasifikasi gangguan tidur yang lebih rendah dibanding ReLU.
